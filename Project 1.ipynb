```python

# Function to identify potential outliers

def identify_outliers(data, column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    return {
        'Column': column,
        'Lower_Bound': lower_bound,
        'Upper_Bound': upper_bound,
        'Outlier_Count': len(outliers),
        'Outlier_Percentage': (
            len(outliers) / len(data) * 100
        )
    }


# Analyze each direct-care staffing category

outlier_results = []

for col in direct_care_columns:
    outlier_results.append(
        identify_outliers(df, col)
    )

outlier_summary = pd.DataFrame(outlier_results)

outlier_summary
```

```python
# Create flags for records that need attention
analysis_df['Zero_Staffing_Flag'] = (
    analysis_df['Total_Direct_Care_Hours'] == 0
).astype(int)

analysis_df['Positive_Census_Flag'] = (
    analysis_df['MDScensus'] > 0
).astype(int)

analysis_df['Zero_Census_Flag'] = (
    analysis_df['MDScensus'] == 0
).astype(int)


# Aggregate data by state
state_summary = analysis_df.groupby('STATE').agg(
    Number_of_Facilities=('PROVNUM', 'nunique'),
    Facility_Days=('PROVNUM', 'count'),
    Total_Resident_Days=('MDScensus', 'sum'),
    Total_Direct_Care_Hours=(
        'Adjusted_Direct_Care_Hours', 'sum'
    ),
    Total_Contract_Hours=(
        'Adjusted_Contract_Hours', 'sum'
    ),
    Zero_Staffing_Days=('Zero_Staffing_Flag', 'sum'),
    Positive_Census_Days=('Positive_Census_Flag', 'sum'),
    Zero_Census_Days=('Zero_Census_Flag', 'sum')
).reset_index()


# Calculate weighted staffing metrics
state_summary['Hours_Per_Resident_Day'] = (
    state_summary['Total_Direct_Care_Hours'] /
    state_summary['Total_Resident_Days'].replace(0, np.nan)
)

state_summary['Contract_Staffing_Percentage'] = (
    state_summary['Total_Contract_Hours'] /
    state_summary['Total_Direct_Care_Hours'].replace(0, np.nan)
) * 100


# Round results
state_summary = state_summary.round(2)


# Sort from highest to lowest staffing hours per resident
state_summary = state_summary.sort_values(
    'Hours_Per_Resident_Day',
    ascending=False
)

display(state_summary.head(10))
```
